```python
import pandas as pd
import numpy as np
import json
import ast
from datetime import datetime
from dateutil.parser import parse
```

```python
# Загружаем файл по итогам проверки результатов нейросети

df = pd.read_excel('./Отчет по нейросети/stats result.xlsx')

cols = ['started_at', 'uuid', 'response', 'raw json', 'name', 'sku (prediction)', 'errorCode', 'source',
        'corr sku', 'value', 'comment']

df = df[cols]
df.rename(columns = {'comment': 'comment_'}, inplace=True)
df['comment'] = None

#Обработка данных и формирование нового файла

for key, row in df.iterrows():
    resp = ast.literal_eval(row.response)
    date = parse(row['started_at'])
    df.loc[df.index == key, 'date'] = datetime.date(date)
    
df.drop('started_at', axis =1, inplace = True)
df.to_excel('./Отчет по нейросети/stats_for_BI_tmp.xlsx', index=False)    
```
